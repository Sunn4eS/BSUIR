## Реализация метода Монте-Карло для вычисления числа π

1. Код с использованием библиотеки `random`
2. Векторизованный вариант через `numpy`

```python
import random


def estimate_pi_monte_carlo(num_points):
    """Вычисляет число Pi методом Монте-Карло.

    num_points: общее количество случайных точек (чем больше, тем точнее)
    """
    points_inside_circle = 0

    for _ in range(num_points):
        # Генерируем случайные координаты x и y в диапазоне [0.0, 1.0)
        x = random.random()
        y = random.random()

        # Проверяем, лежит ли точка внутри четверти круга радиуса 1
        # Формула: x^2 + y^2 <= 1^2
        if x**2 + y**2 <= 1.0:
            points_inside_circle += 1

    # Отношение площадей (pi * r^2 / 4) / r^2  равно pi / 4.
    # Следовательно, pi = 4 * (кол-во точек в круге / всего точек)
    pi_estimate = 4 * points_inside_circle / num_points

    return pi_estimate


# --- Пример использования ---
# 100 тысяч точек обычно дают точность до 2-3 знаков после запятой
total_points = 10_000_000
calculated_pi = estimate_pi_monte_carlo(total_points)

print(f"Количество симулированных точек: {total_points:,}")
print(f"Приближенное значение Pi: {calculated_pi}")
print(f"Абсолютная погрешность: {abs(3.1415926535 - calculated_pi):.6f}")
```

```python
import numpy as np


def estimate_pi_numpy(num_points):
    # Генерируем сразу матрицу случайных координат (num_points строк, 2 колонки)
    points = np.random.rand(num_points, 2)

    # Возводим координаты в квадрат и складываем x^2 + y^2 для каждой строки
    distances = np.sum(points**2, axis=1)

    # Считаем, сколько сумм меньше или равны 1
    points_inside_circle = np.sum(distances <= 1.0)

    # Вычисляем Pi
    return 4 * points_inside_circle / num_points


# 10 миллионов точек обработаются меньше чем за секунду
calculated_pi = estimate_pi_numpy(10_000_000)
print(f"Быстрый расчет Pi (NumPy): {calculated_pi}")
print(f"Абсолютная погрешность: {abs(3.1415926535 - calculated_pi):.6f}")
```
